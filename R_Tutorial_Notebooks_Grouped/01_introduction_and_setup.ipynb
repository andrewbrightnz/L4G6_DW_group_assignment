## 1.1 What is R?

R is a programming language built specifically for statistics, data analysis,
and graphics, created in the early 1990s by Ross Ihaka and Robert Gentleman
at the University of Auckland. Alongside Python, it's one of the two
dominant languages in data science and statistics.

**Why people reach for R:**
- A huge ecosystem of statistical and plotting functions built in
- Thousands of add-on packages (CRAN) for almost any analytical task
- Strong, mature visualisation tools (base graphics, `ggplot2`, etc.)
- Free, open source, cross-platform (Windows, macOS, Linux)

**Common uses:** academic research and statistics, bioinformatics, finance
and econometrics, data journalism, general data wrangling and machine
learning.


```r
# A quick taste of R - descriptive statistics on a numeric vector
scores <- c(88, 92, 79, 95, 85, 90, 76, 98)

cat("Mean:  ", mean(scores), "\n")
cat("Median:", median(scores), "\n")
cat("SD:    ", sd(scores), "\n")
```

## 1.2 Getting Started

To run R locally you typically install two things:

1. **R itself** - from CRAN (https://cran.r-project.org).
2. **An editor/IDE** - most people use **RStudio**. You can also run R
   through Jupyter with the `IRkernel` package (as this notebook does), or

**A minimal workflow:** write code -> run it -> inspect the output.


```r
# The classic first program
print("Hello, World!")
```

```r
# In an interactive session or notebook, a bare variable name auto-prints
message <- "Hello, World!"
message
```

## 1.3 Syntax

R programs are made of expressions - assignments, function calls, and
control-flow statements.

- Assignment is idiomatically written with `<-` (though `=` also works).
- Names are case-sensitive (`Age` and `age` are different).
- Code blocks (`if`, `for`, functions, etc.) use curly braces `{ }`.
- Comments start with `#` and run to the end of the line.


```r
x <- 10
y <- 3
total <- x + y
total
```

```r
if (total > 10) {
  print("total is greater than 10")
}
```

## 1.4 Comments

R only supports single-line comments (no dedicated multi-line syntax) -
everything after `#` on a line is ignored.


```r
# This whole line is a comment and does nothing
x <- 5  # comments can also trail after code on the same line
x
```

```r
# A "block" comment is really just several single-line comments in a row.
# y <- 10   <- commented out; this line does not run
y <- 20
y
```
