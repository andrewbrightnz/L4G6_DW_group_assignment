# 2. Variables

This notebook covers: **creating variables, printing them, concatenating
values into strings, working with several variables at once, and the rules
for naming them.**


## 2.1 Creating Variables

A variable is a named container for a value, created simply by assigning a
value to a name with `<-`. There's no separate declaration step, and no
fixed type attached to the name itself - only the value has a type.


```r
name <- "Andrew"
age  <- 29
city <- "Christchurch"

name
age
city
```

```r
# Because variables aren't typed, a name can be reassigned to a different
# type later
value <- 100        # numeric
value <- "hundred"  # now a character string - perfectly legal
value
```

## 2.2 Printing Variables

There are three common ways to display a variable's value:
1. Type the name alone at the top level - R auto-prints it.
2. Call `print()` explicitly.
3. Use `cat()` to print without the `[1]` index prefix.


```r
city <- "Christchurch"

city             # auto-print
print(city)      # explicit print
cat(city, "\n")   # cat() - no quotes, no index prefix
```

```r
# The difference is clearer with vectors - print() shows an index like [1]
numbers <- c(10, 20, 30, 40, 50)    # c() builds the vector
print(numbers)
cat(numbers, "\n")  # cat() concatenates values into text and prints them — it's for display/output only.
```

## 2.3 Concatenating Variables

"Concatenating" means joining pieces of text together.

- `paste()` joins values with a separator (a space, by default).
- `paste0()` is shorthand for `paste(..., sep = "")` - no separator.


```r
first <- "Andrew"
last  <- "Smith"

full_name <- paste(first, last)
full_name
```

```r
tag <- paste0("user_", 42)      # no space between the pieces
tag
```

```r
# paste() coerces numbers to strings automatically
sentence <- paste("There are", 7, "days in a week")
sentence
```

## 2.4 Multiple Variables

R lets you assign several variables on one line, or chain the same value to
several names at once.


```r
a <- 1; b <- 2; c <- 3
a
b
c
```

```r
x <- y <- z <- "same value"   # chained assignment
x
y
z
```

```r
# R has no built-in destructuring, but you can unpack a vector manually
values <- c(10, 20, 30)
p <- values[1]  # In R, the index starts at 1, not 0 like in Python
q <- values[2]
r <- values[3]

p
q
r
```

## 2.5 Variable Naming Rules

**Must follow:**
- Names can contain letters, digits, `.` and `_`.
- A name cannot start with a digit or `_`.
- A name starting with `.` cannot be immediately followed by a digit.
- Reserved words (`if`, `else`, `for`, `function`, `TRUE`, `FALSE`, `NULL`,
  `NA`, etc.) can't be used as names.

**Convention (not enforced):** `snake_case` or `camelCase`, chosen
consistently, with descriptive names.


```r
my_var <- 1
my.var <- 2
MYVAR  <- 3
myvar2 <- 4

my_var; my.var; MYVAR; myvar2
```

```r
# Swapping the values of two variables without a third helper name
p1 <- "left"
p2 <- "right"

temp <- p1
p1 <- p2
p2 <- temp

p1
p2
```
