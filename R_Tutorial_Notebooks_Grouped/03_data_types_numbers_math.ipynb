# 3. Data Types, Numbers & Math

This notebook covers: **R's basic data types, the numeric types specifically,
and the built-in math functions.**


## 3.1 Data Types

Every value in R has an underlying type. The most common basic types:

| Type        | Example        | Description                          |
|-------------|----------------|---------------------------------------|
| `numeric`   | `4.5`          | Any real number (default for numbers) |
| `integer`   | `4L`           | Whole numbers (`L` suffix forces integer storage) |
| `complex`   | `2+3i`         | Complex numbers                       |
| `character` | `"hello"`      | Text strings                          |
| `logical`   | `TRUE`/`FALSE` | Boolean values                        |

Check a value's type with `class()` or `typeof()`.


```r
num  <- 42.5
int  <- 42L     # L tells R the number is an integer, not a double
comp <- 3 + 2i  # i tells R the number is an imaginary number
chr  <- "R programming"
lgl  <- TRUE

cat("class(num): ", class(num),  "\n")
cat("class(int): ", class(int),  "\n")
cat("class(comp):", class(comp), "\n")
cat("class(chr): ", class(chr),  "\n")
cat("class(lgl): ", class(lgl),  "\n")
```

```r
typeof(42)   # "double" - a plain number defaults to numeric, not integer
typeof(42L)  # "integer" - explicitly forced with the L suffix
```

## 3.2 Numbers

Most everyday arithmetic uses `numeric` (double-precision). Very large or
small numbers can be written in scientific notation with `e`.


```r
x <- 10       # numeric (double) by default
y <- 10L      # explicitly an integer
z <- 10.5     # numeric with a decimal component

class(x); class(y); class(z)
```

'e' is scientific notation for 'exponent' or 'to the power of'

e.g. 2e3 means 2 cubed

```r
big_number   <- 5e6    # 5,000,000
small_number <- 5e-3   # 0.005

big_number
small_number
```

```r
a <- 15
b <- 4

a + b   # addition
a - b   # subtraction
a * b   # multiplication
a / b   # division
a %% b  # modulus (remainder)
a %/% b # integer division
```

## 3.3 Math Functions

R has a large built-in math library on top of arithmetic operators.


```r
values <- c(4, 9, 25, -7, 3.7)

max(values)
min(values)
sqrt(values[values > 0])    # The function in the square brackets filters the results 
                            # to only those that meet the function (values greater than 0)
abs(values)
```

```r
n <- 7.456

round(n)      # nearest whole number
round(n, 1)   # nearest tenth/1 decimal place
round(n, 3)   # nearest tenth/3 decimal places
ceiling(n)    # rounds up
floor(n)      # rounds down
```

```r
log(100)      # natural log
log10(100)    # base-10 log
exp(1)        # e^1
factorial(5)  # 5!
```

## More examples


```r
# Checking a value's type before doing arithmetic on it
mystery <- "42"
is.numeric(mystery)          # FALSE - it's a string
mystery_num <- as.numeric(mystery)  # Converts mystery to a number
is.numeric(mystery_num)      # TRUE after conversion
mystery_num + 8
```

```r
# Vectorised math - functions apply element-by-element automatically
temps_celsius <- c(0, 10, 20, 30, 40)
temps_fahrenheit <- temps_celsius * 9/5 + 32
temps_fahrenheit
```

```r
# Rounding a whole vector of prices to 2 decimal places
prices <- c(19.996, 4.005, 100.5555)
round(prices, 2)
```
