# 4. Strings

This notebook covers: **creating and inspecting strings, and using escape
characters for special content like quotes, newlines, and tabs.**


## 4.1 Strings

Strings are written with single or double quotes and are of type
`character`. R has a rich set of built-in functions for working with text.


```r
greeting <- "Hello there"
class(greeting)
nchar(greeting)      # length in characters
```

```r
toupper(greeting)
tolower(greeting)
```

```r
substr(greeting, 1, 5)   # characters 1 through 5
```

```r
sub("there", "world", greeting)     # replace first match
gsub("l", "L", greeting)            # replace all matches
```

## 4.2 Escape Characters

To include a literal quote (or other awkward character) inside a string,
use a backslash `\` escape.

| Sequence | Meaning        |
|----------|----------------|
| `\"`     | double quote   |
| `\'`     | single quote   |
| `\\`     | backslash      |
| `\n`     | new line       |
| `\t`     | tab            |


```r
quote_example <- "She said \"R is great\" to me."
cat(quote_example, "\n")
```

```r
multiline <- "First line\nSecond line"
cat(multiline, "\n")
```

```r
tabbed <- "Name:\tAndrew"
cat(tabbed, "\n")
```

## More examples


```r
# Splitting a string into pieces
sentence <- "R is fun to learn"
words <- strsplit(sentence, " ")[[1]]
words
length(words)
```

```r
# Checking whether a string contains a substring
grepl("fun", sentence)
grepl("boring", sentence)
```

```r
# Padding and trimming whitespace
padded <- "   trim me   "
trimws(padded)

formatC("7", width = 3, flag = "0")   # pad a number-as-string with zeros
```

```r
# Building a formatted report line from several pieces of data
item  <- "Notebook"
price <- 4.5
qty   <- 3

cat(sprintf("%-10s x%d = $%.2f\n", item, qty, price * qty))
```
