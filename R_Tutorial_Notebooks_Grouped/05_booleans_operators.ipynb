# 5. Booleans & Operators

This notebook covers: **logical (TRUE/FALSE) values, and R's arithmetic,
comparison, logical, and membership operators.**


## 5.1 Booleans

The logical type has exactly two values: `TRUE` and `FALSE` (shorthand `T`/
`F` also work, but the full words are safer since `T`/`F` can be reassigned
as ordinary variables). Booleans commonly come from comparisons, and `1`/`0`
coerce to `TRUE`/`FALSE`.


```r
is_student <- TRUE
has_graduated <- FALSE

is_student
class(is_student)
```

```r
10 > 5    # TRUE
10 == 5   # FALSE
10 != 5   # TRUE
```

```r
as.logical(1)   # TRUE
as.logical(0)   # FALSE
```

## 5.2 Operators

- **Arithmetic:** `+ - * / ^ %% %/%`
- **Assignment:** `<-`, `<<-`, `=`, `->`
- **Comparison:** `== != > < >= <=`
- **Logical:** `& | ! && ||`
- **Miscellaneous:** `:` (sequence), `%in%` (membership)


```r
5 + 2; 5 - 2; 5 * 2; 5 / 2; 5 ^ 2; 5 %% 2; 5 %/% 2
```

```r
7 == 7
7 != 8
7 > 3
```

```r
# & and | work element-wise on vectors; && and || short-circuit on scalars
c(TRUE, FALSE) & c(TRUE, TRUE)
TRUE && FALSE
```

```r
"b" %in% c("a", "b", "c")
```

## More examples


```r
# Combining comparisons with logical operators
age <- 22
has_ticket <- TRUE

can_enter <- age >= 18 & has_ticket
can_enter
```

```r
# ! negates a logical value
is_weekday <- TRUE
!is_weekday
```

```r
# %in% is handy for filtering a vector against a set of allowed values
allowed_grades <- c("A", "B", "C")
student_grades <- c("A", "F", "B", "D", "C")

student_grades %in% allowed_grades
student_grades[student_grades %in% allowed_grades]
```

```r
# The : operator quickly builds an integer sequence, useful in comparisons
evens <- (1:10)[(1:10) %% 2 == 0]
evens
```
