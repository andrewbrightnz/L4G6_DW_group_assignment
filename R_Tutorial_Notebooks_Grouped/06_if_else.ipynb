# 6. If...Else

This notebook covers: **basic if/else branching, the vectorised `ifelse()`,
and nested conditionals.**


## 6.1 If...Else

`if`/`else` branches based on a condition, which must evaluate to a single
`TRUE` or `FALSE`.


```r
temperature <- 15

if (temperature > 25) {
  print("It's warm")
} else if (temperature > 10) {
  print("It's mild")
} else {
  print("It's cold")
}
```

`ifelse()` is a compact, **vectorised** alternative - useful for applying a
condition across an entire vector at once, rather than looping.


```r
temps <- c(5, 12, 22, 30)
ifelse(temps > 20, "warm", "cool")
```

## 6.2 Nested If...Else

Conditions can be nested when a decision has more than one level of
branching. When there are several mutually exclusive options, chaining
`else if` is usually more readable than deep nesting.


```r
age <- 20
has_id <- TRUE

if (age >= 18) {
  if (has_id) {
    print("Entry allowed")
  } else {
    print("ID required")
  }
} else {
  print("Too young to enter")
}
```

```r
grade <- 82

if (grade >= 90) {
  result <- "A"
} else if (grade >= 80) {
  result <- "B"
} else if (grade >= 70) {
  result <- "C"
} else {
  result <- "F"
}

result
```

## More examples


```r
# switch() is a tidy alternative to a long else-if chain for exact matches
day <- "Wed"

day_type <- switch(day,
  Sat = ,
  Sun = "Weekend",
  "Weekday")

day_type
```

```r
# Grading a whole vector of scores at once with ifelse()
scores <- c(95, 82, 71, 58, 90)
grades <- ifelse(scores >= 90, "A",
           ifelse(scores >= 80, "B",
            ifelse(scores >= 70, "C", "F")))
grades
```

```r
# Nested condition combined with vector filtering
numbers <- 1:20
divisible_by_3_and_5 <- numbers[numbers %% 3 == 0 & numbers %% 5 == 0]
divisible_by_3_and_5
```
