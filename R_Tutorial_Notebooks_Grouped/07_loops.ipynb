# 7. Loops

This notebook covers: **`while` loops, `for` loops, and the `break`/`next`
control statements.**


## 7.1 While Loop

A `while` loop repeats a block as long as its condition stays `TRUE`. It's
the right tool when you don't know in advance how many iterations you need.


```r
i <- 1
while (i <= 5) {
  print(i)
  i <- i + 1
}
```

`break` exits a loop immediately; `next` skips to the following iteration
without finishing the current one.


```r
i <- 0
while (i < 10) {
  i <- i + 1
  if (i %% 2 == 0) {
    next   # skip even numbers
  }
  if (i > 7) {
    break  # stop once i exceeds 7
  }
  print(i)
}
```

## 7.2 For Loop

A `for` loop iterates over the elements of a sequence - a vector, list, or
range - running the loop body once per element.


```r
fruits <- c("apple", "banana", "cherry")

for (fruit in fruits) {
  print(fruit)
}
```

```r
# Looping over a numeric range with the : operator
for (i in 1:5) {
  print(i * i)
}
```

```r
# break and next work the same way inside a for loop
for (i in 1:10) {
  if (i == 6) {
    break
  }
  print(i)
}
```

## More examples


```r
# Accumulating a result across iterations - a running total
total <- 0
for (n in 1:10) {
  total <- total + n
}
total
```

```r
# Looping over a list of named items
prices <- list(apple = 0.5, bread = 2.3, milk = 1.8)

for (item_name in names(prices)) {
  cat(item_name, "costs $", prices[[item_name]], "\n", sep = "")
}
```

```r
# Nested loops - a simple multiplication table
for (i in 1:3) {
  for (j in 1:3) {
    cat(i, "x", j, "=", i * j, "  ")
  }
  cat("\n")
}
```

```r
# Preferring a vectorised approach over a loop where possible (often faster
# and more idiomatic in R)
squares_loop <- numeric(5)
for (i in 1:5) {
  squares_loop[i] <- i^2
}
squares_loop

squares_vectorised <- (1:5)^2   # same result, no explicit loop needed
squares_vectorised
```
