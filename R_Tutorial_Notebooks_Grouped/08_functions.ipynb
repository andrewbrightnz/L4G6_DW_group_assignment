# 8. Functions, Parameters & Scope

This notebook covers: **defining functions, parameters/arguments (including
defaults and named arguments), and variable scope.**


## 8.1 Functions

A function is a reusable, named block of code, defined with `function()` and
assigned to a name - functions are first-class objects in R.


```r
greet <- function() {
  print("Hello from a function!")
}

greet()
```

```r
square <- function(x) {
  return(x^2)
}

square(6)
```

If there's no explicit `return()`, R returns the value of the last evaluated
expression in the function body.


```r
cube <- function(x) {
  x^3   # no return() needed
}

cube(3)
```

## 8.2 Parameters and Arguments

A **parameter** is the placeholder name in the function definition; an
**argument** is the actual value passed in a call. R supports default
parameter values and named arguments.


```r
introduce <- function(name, age) {
  cat(name, "is", age, "years old\n")
}

introduce("Andrew", 29)
```

```r
introduce_default <- function(name, age = 18) {
  cat(name, "is", age, "years old\n")
}

introduce_default("Sam")           # uses the default age
introduce_default("Sam", 25)       # overrides the default
```

```r
introduce_default(age = 40, name = "Priya")   # named arguments, any order
```

```r
# ... lets a function accept a variable number of arguments
sum_all <- function(...) {
  sum(...)
}

sum_all(1, 2, 3, 4)
```

## 8.3 Variable Scope

- A variable created *inside* a function is **local**: it disappears once
  the function finishes.
- A variable created *outside* any function is **global** and can be read
  from inside functions (but not modified directly, by default).
- `<<-` ("super assignment") lets a function modify an enclosing/global
  variable.


```r
x <- "global value"

show_x <- function() {
  print(x)   # reads the global x fine
}

show_x()
```

```r
modify_local <- function() {
  x <- "local value"   # creates a NEW local variable
  print(x)
}

modify_local()
print(x)   # unchanged
```

```r
modify_global <- function() {
  x <<- "changed by <<-"
}

modify_global()
print(x)
```

## More examples


```r
# Functions can return more than one value by bundling them into a list
summary_stats <- function(x) {
  list(mean = mean(x), sd = sd(x), n = length(x))
}

result <- summary_stats(c(4, 8, 15, 16, 23, 42))
result$mean
result$sd
```

```r
# Recursive functions - a function that calls itself
factorial_recursive <- function(n) {
  if (n <= 1) {
    return(1)
  }
  n * factorial_recursive(n - 1)
}

factorial_recursive(5)
```

```r
# Functions as arguments to other functions (R treats them as regular values)
apply_twice <- function(f, x) {
  f(f(x))
}

apply_twice(function(x) x * 2, 3)   # (3*2)*2 = 12
```

```r
# sapply()/lapply() apply a function across every element of a vector/list
# without writing an explicit for loop
nums <- c(1, 2, 3, 4, 5)
sapply(nums, function(x) x^2)
```
