# 9. Vectors & Lists

This notebook covers: **vectors (same-type collections) and lists
(mixed-type collections), including indexing, slicing, and modification.**


## 9.1 Vectors

A vector is R's most fundamental data structure: an ordered collection of
values all of the same type, created with `c()` ("combine").


```r
fruits <- c("banana", "apple", "orange")
numbers <- c(1, 2, 3, 4, 5)

fruits
numbers
class(fruits)
class(numbers)
```

Elements are accessed with square brackets - **indexing starts at 1**, not 0.


```r
fruits[1]              # first element
fruits[2:3]            # a slice - elements 2 through 3
numbers[numbers > 2]   # elements satisfying a condition
```

```r
seq(from = 2, to = 20, by = 2)   # a regular sequence

a <- c(1, 2, 3)
b <- c(10, 20, 30)
a + b   # arithmetic on vectors happens element-wise
```

## 9.2 Lists

Unlike a vector, a **list** can hold elements of different types - numbers,
strings, vectors, even other lists - created with `list()`.


```r
person <- list("Andrew", 29, TRUE)
person
```

Named list elements make access clearer: use `[[ ]]` for a single element or
`$` for access by name.


```r
person <- list(name = "Andrew", age = 29, student = TRUE)

person$name
person[["age"]]
```

```r
length(person)
"student" %in% names(person)

person$city <- "Christchurch"   # add a new named element
person
```

```r
person$student <- NULL   # remove an element by setting it to NULL
person
```

## More examples


```r
# Vector recycling - the shorter vector is repeated to match the longer one
c(1, 2, 3, 4, 5, 6) + c(10, 100)
```

```r
# Negative indices exclude elements rather than select them
letters_sample <- c("a", "b", "c", "d", "e")
letters_sample[-1]      # everything except the first element
letters_sample[-c(1,2)] # everything except the first two
```

```r
# sorting and reversing a vector
scores <- c(72, 91, 60, 85, 77)
sort(scores)
sort(scores, decreasing = TRUE)
rev(scores)
```

```r
# A list of lists - nested structures for more complex data
students <- list(
  list(name = "Andrew", grade = "B"),
  list(name = "Priya", grade = "A")
)

students[[2]]$name
sapply(students, function(s) s$grade)
```
