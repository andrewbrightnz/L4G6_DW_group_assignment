# 10. Matrices & Arrays

This notebook covers: **two-dimensional matrices and their generalisation to
N dimensions, arrays.**


## 10.1 Matrices

A matrix is a two-dimensional structure where every element shares the same
type - a vector reshaped into rows and columns, built with `matrix()`.


```r
m <- matrix(1:6, nrow = 2, ncol = 3)
m
```

By default `matrix()` fills column by column; `byrow = TRUE` fills row by
row instead. Cells, rows, or columns are accessed with `[row, col]`.


```r
m2 <- matrix(1:6, nrow = 2, byrow = TRUE)
m2

m2[1, 3]     # row 1, column 3
m2[2, ]      # entire second row
m2[, 1]      # entire first column
```

```r
# Matrix arithmetic is element-wise unless you ask for true matrix math
a <- matrix(1:4, nrow = 2)
b <- matrix(5:8, nrow = 2)

a + b        # element-wise addition
a %*% b      # true matrix multiplication
```

## 10.2 Arrays

While a matrix is limited to two dimensions, an **array** generalises to any
number of dimensions, via `array()` with a `dim` vector describing the
shape.


```r
arr <- array(1:12, dim = c(2, 3, 2))   # 2 rows, 3 cols, 2 "layers"
arr
```

```r
arr[1, 2, 2]   # a single element via three indices
arr[, , 1]     # an entire "layer"
```

```r
dim(arr)   # reports the shape of the array
```

## More examples


```r
# Naming rows and columns makes matrices much easier to read
scores <- matrix(c(88, 92, 79, 95), nrow = 2,
                  dimnames = list(c("Andrew", "Priya"), c("Test1", "Test2")))
scores

scores["Priya", "Test1"]
```

```r
# rowSums / colSums / rowMeans / colMeans are quick aggregate helpers
grades <- matrix(c(85, 90, 78, 92, 88, 95), nrow = 2, byrow = TRUE)
rowSums(grades)
colMeans(grades)
```

```r
# Transposing a matrix - flipping rows and columns
t(grades)
```

```r
# Combining matrices with rbind() and cbind()
extra_row <- c(70, 75, 80)
rbind(grades, extra_row)

extra_col <- c(1, 2)
cbind(grades, extra_col)
```
