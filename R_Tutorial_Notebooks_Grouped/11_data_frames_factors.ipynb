# 11. Data Frames & Factors

This notebook covers: **data frames (R's table-like structure for real
data) and factors (categorical data).**


## 11.1 Data Frames

A data frame is R's structure for tabular data - conceptually a spreadsheet
or SQL table. Each column can hold a different type, but every column must
have the same length. Built with `data.frame()`, it's the workhorse
structure for real-world analysis in R.


```r
students <- data.frame(
  name = c("Andrew", "Priya", "Sam"),
  age  = c(29, 24, 31),
  pass = c(TRUE, TRUE, FALSE)
)

students
```

```r
nrow(students)
ncol(students)
str(students)
```

```r
students$name                # a single column, by name
students[1, ]                # the first row
students[students$pass, ]    # rows where pass is TRUE
```

```r
students$year <- c(1, 2, 3)   # adding a new column
students
```

## 11.2 Factors

A factor represents **categorical** data - a fixed, limited set of possible
values (levels), like "Low"/"Medium"/"High". Internally the categories are
stored as integers, with text labels displayed - more memory-efficient than
repeated strings.


```r
sizes <- factor(c("small", "medium", "large", "medium", "small"))
sizes
levels(sizes)
```

By default levels are ordered alphabetically with no inherent ranking. For a
naturally ordered variable, tell R about the order explicitly.


```r
sizes_ordered <- factor(
  c("small", "medium", "large", "medium", "small"),
  levels = c("small", "medium", "large"),
  ordered = TRUE
)

sizes_ordered
sizes_ordered[1] < sizes_ordered[3]   # TRUE, since small < large
```

```r
summary(sizes)   # counts of each level
```

## More examples


```r
# Using a factor as a column inside a data frame
survey <- data.frame(
  respondent = c("A", "B", "C", "D"),
  satisfaction = factor(c("high", "low", "medium", "high"),
                         levels = c("low", "medium", "high"), ordered = TRUE)
)

survey
survey[survey$satisfaction >= "medium", ]
```

```r
# order() to sort a data frame by a column
students <- data.frame(
  name  = c("Andrew", "Priya", "Sam"),
  score = c(88, 95, 76)
)

students[order(students$score, decreasing = TRUE), ]
```

```r
# apply()-style summaries across a data frame's numeric columns
grades <- data.frame(
  test1 = c(85, 90, 78),
  test2 = c(92, 88, 95)
)

sapply(grades, mean)
```

```r
# merging two data frames (like a SQL join) with merge()
contact_info <- data.frame(name = c("Andrew", "Priya"), email = c("a@x.com", "p@x.com"))
scores_info  <- data.frame(name = c("Andrew", "Priya"), score = c(88, 95))

merge(contact_info, scores_info, by = "name")
```
