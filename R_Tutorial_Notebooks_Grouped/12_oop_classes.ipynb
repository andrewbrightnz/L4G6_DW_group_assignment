# 12. OOP: Classes and Objects

This notebook covers: **object-oriented programming in R, focusing on the
two most common systems, S3 and S4.**


## 12.1 S3 Classes

R supports several competing OOP systems. **S3** is informal and
lightweight: any object can be given a `class` attribute, and functions can
be written to behave differently depending on that class ("generic"
functions with method dispatch).


```r
# S3: build an object by attaching a class attribute to a list
car <- list(make = "Toyota", model = "Corolla", year = 2020)
class(car) <- "Car"

car
class(car)
```

```r
# Define a generic and a method for the "Car" class
print.Car <- function(x, ...) {
  cat("Car:", x$year, x$make, x$model, "\n")
}

print(car)   # dispatches to print.Car automatically
```

## 12.2 S4 Classes

**S4** is more formal: classes and their fields ("slots") are defined
explicitly with `setClass()`, which brings stronger type-checking than S3.


```r
setClass("Person", representation(name = "character", age = "numeric"))

p <- new("Person", name = "Andrew", age = 29)
p@name
p@age
```

```r
# S4 methods are also defined explicitly, via setGeneric()/setMethod()
setGeneric("describe", function(x) standardGeneric("describe"))

setMethod("describe", "Person", function(x) {
  cat(x@name, "is", x@age, "years old\n")
})

describe(p)
```

## More examples


```r
# S3: a second class showing method dispatch choosing different behaviour
summary.Car <- function(object, ...) {
  cat("A", object$year, object$make, object$model, "\n")
}

summary(car)
```

```r
# S3 inheritance - a class can extend another via a class vector
sports_car <- list(make = "Mazda", model = "MX-5", year = 2022, top_speed = 217)
class(sports_car) <- c("SportsCar", "Car")   # inherits from "Car"

print(sports_car)   # falls back to print.Car since no print.SportsCar exists
```

```r
# S4: validity checks can enforce rules on slot values
setClass("Rectangle", representation(width = "numeric", height = "numeric"),
  validity = function(object) {
    if (object@width <= 0 || object@height <= 0) {
      "width and height must be positive"
    } else {
      TRUE
    }
  })

r <- new("Rectangle", width = 4, height = 3)
r@width * r@height   # area
```

```r
# Checking whether an object belongs to a given class
is(p, "Person")
inherits(sports_car, "Car")
```
