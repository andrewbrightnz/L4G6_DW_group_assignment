## 13.1 Packages

A package bundles functions, data, and documentation for reuse. R ships with
"base" packages already loaded, but its real power comes from ~22,000
add-on packages on CRAN and elsewhere.

**Typical workflow:**
1. `install.packages("packagename")` - once per machine.
2. `library(packagename)` - once per session that uses it.


```r
# install.packages("dplyr")  # commented out - requires network access

library(stats)   # a base package, always available
"stats package loaded"
```

```r
(.packages())   # packages currently loaded in this session
```

## 13.2 Built-in Data Sets

```r
head(iris)   # classic data set - measurements of iris flowers
```

## 13.3 Reading and Writing CSV Files

Real analysis usually starts by loading data from a file. Base R's
`read.csv()` and `write.csv()` handle the most common tabular format.


```r
df <- data.frame(
  name  = c("Andrew", "Priya", "Sam"),
  score = c(88, 95, 76)
)

write.csv(df, "students.csv", row.names = FALSE)
list.files(pattern = "\\.csv$")
```

```r
loaded <- read.csv("students.csv")
loaded
str(loaded)
```

For very large files, `data.table` (`fread()`/`fwrite()`) or `readr`
(`read_csv()`/`write_csv()`) are considerably faster than the base
equivalents.


## More examples


```r
# Filtering and summarising a built-in dataset without any extra packages
setosa_only <- iris[iris$Species == "setosa", ]
mean(setosa_only$Sepal.Length)
```

```r
# Appending new rows to an existing CSV-backed data frame, then re-saving
new_student <- data.frame(name = "Priya", score = 91)
updated <- rbind(loaded, new_student)
write.csv(updated, "students.csv", row.names = FALSE)
read.csv("students.csv")
```
