# 14. Descriptive Statistics

This notebook covers: **an introduction to statistics in R, plus the core
descriptive measures - mean/median/mode, percentiles, and standard
deviation/variance.**


## 14.1 Introduction

R was built by statisticians, so descriptive and inferential statistics are
first-class operations. The running example throughout this notebook is a
small data set of recorded car speeds.


```r
speeds <- c(99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86)
speeds
length(speeds)
```

```r
summary(speeds)   # a quick five-number summary plus the mean
```

## 14.2 Mean, Median, and Mode

- **Mean** - the arithmetic average.
- **Median** - the middle value once sorted.
- **Mode** - the most frequent value. R has no built-in statistical `mode()`
  function (the base `mode()` reports storage type instead), so a small
  helper is common.


```r
mean(speeds)
median(speeds)
```

```r
get_mode <- function(x) {
  freq <- table(x)
  names(freq)[freq == max(freq)]
}

get_mode(speeds)
```

## 14.3 Percentiles

A percentile is the value below which a given percentage of the data falls.
The 50th percentile is the median; the 25th/75th mark the interquartile
range. Computed with `quantile()`.


```r
quantile(speeds, 0.75)   # 75th percentile
quantile(speeds, 0.25)   # 25th percentile
```

```r
quantile(speeds)   # several common percentiles at once
```

```r
IQR(speeds)   # interquartile range - the spread of the middle 50%
```

## 14.4 Standard Deviation and Variance

- **Variance** - the average squared difference from the mean.
- **Standard deviation** - the square root of the variance, back in the
  original units, making it easier to interpret.

Low standard deviation = tightly clustered values; high = widely spread.


```r
speeds1 <- c(32, 111, 138, 28, 59, 77, 97)
speeds2 <- c(85, 86, 86, 87, 87, 88, 94)

sd(speeds1)   # widely spread -> larger standard deviation
sd(speeds2)   # tightly clustered -> smaller standard deviation
```

```r
var(speeds1)     # variance
sd(speeds1)^2    # confirms variance = sd squared
```

## More examples


```r
# Comparing two groups' central tendency and spread side by side
group_a <- c(70, 72, 68, 75, 71)
group_b <- c(60, 90, 55, 85, 60)

cat("Group A - mean:", mean(group_a), " sd:", sd(group_a), "\n")
cat("Group B - mean:", mean(group_b), " sd:", sd(group_b), "\n")
```

```r
# Detecting outliers using the 1.5*IQR rule
q1 <- quantile(speeds, 0.25)
q3 <- quantile(speeds, 0.75)
iqr <- q3 - q1

lower_bound <- q1 - 1.5 * iqr
upper_bound <- q3 + 1.5 * iqr

speeds[speeds < lower_bound | speeds > upper_bound]   # any outliers found
```

```r
# boxplot() gives a quick visual summary of median, quartiles, and outliers
boxplot(speeds, main = "Distribution of Car Speeds", ylab = "km/h")
```
