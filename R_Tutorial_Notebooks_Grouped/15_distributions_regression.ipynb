# 15. Distributions, Scatter Plots & Regression

This notebook covers: **generating and visualising a normal distribution,
plotting relationships with scatter plots, and fitting a linear regression
model.**


## 15.1 Normal Data Distribution

The normal ("Gaussian"/bell curve) distribution is the most important
distribution in statistics: values cluster symmetrically around a mean,
becoming rarer further away from it. `rnorm()` generates random samples.


```r
set.seed(42)   # for reproducible random numbers

sample_data <- rnorm(1000, mean = 5, sd = 1)

mean(sample_data)
sd(sample_data)
```

```r
hist(sample_data,
     breaks = 30,
     main = "Histogram of a Normal Distribution",
     xlab = "Value",
     col = "steelblue")
```

`dnorm()`, `pnorm()`, and `qnorm()` give the density, cumulative
probability, and quantile functions respectively.


```r
pnorm(6, mean = 5, sd = 1)   # probability that a value is <= 6
```

## 15.2 Scatter Plots

A scatter plot shows the relationship between two numeric variables, one
point per observation - the standard first check for correlation before
fitting any model.


```r
car_age <- c(5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6)
car_speed <- c(99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86)

plot(car_age, car_speed,
     main = "Car Age vs Speed",
     xlab = "Age (years)",
     ylab = "Speed (km/h)",
     pch = 19,
     col = "darkred")
```

## 15.3 Linear Regression

Linear regression fits a straight line that best explains the relationship
between an explanatory variable (x) and a response variable (y), via `lm()`.


```r
model <- lm(car_speed ~ car_age)
summary(model)
```

`(Intercept)` is where the fitted line crosses the y-axis; the `car_age`
coefficient is the slope. `Multiple R-squared` shows how much of the
variation in speed is explained by age alone.


```r
plot(car_age, car_speed,
     main = "Linear Regression: Car Age vs Speed",
     xlab = "Age (years)",
     ylab = "Speed (km/h)",
     pch = 19,
     col = "darkred")
abline(model, col = "blue", lwd = 2)
```

```r
new_age <- data.frame(car_age = 10)
predict(model, new_age)   # using the model to predict for a new value
```

## More examples


```r
# Checking correlation strength directly with cor()
cor(car_age, car_speed)   # close to -1 = strong negative correlation
```

```r
# Multiple regression - predicting from more than one explanatory variable
set.seed(1)
engine_size <- car_age * 0.1 + rnorm(length(car_age), sd = 0.3)

multi_model <- lm(car_speed ~ car_age + engine_size)
summary(multi_model)
```

```r
# Visualising residuals - the gaps between actual and predicted values
plot(model$fitted.values, model$residuals,
     main = "Residuals vs Fitted",
     xlab = "Fitted values", ylab = "Residuals",
     pch = 19, col = "darkgreen")
abline(h = 0, lty = 2)
```

```r
# Comparing several random samples drawn from different distributions
set.seed(7)
normal_sample <- rnorm(500, mean = 0, sd = 1)
uniform_sample <- runif(500, min = -3, max = 3)

par(mfrow = c(1, 2))
hist(normal_sample, main = "Normal", col = "steelblue", breaks = 20)
hist(uniform_sample, main = "Uniform", col = "orange", breaks = 20)
par(mfrow = c(1, 1))
```
